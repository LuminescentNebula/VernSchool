# Деревья
Списки и словари размещают данные линейно, но далеко не все реальные данные возможно разместить просто друг за другом. Для представления иерархических данных используются иерархические структуры данных, например дерево.

Дерево - связанный граф.

Дерево состоит из корневого узла - самого верхнего узла в графе, у которого нет родителя, внутренний узлов и листовых узлов, у которых нет родителей.

В Дереве как в графе нет циклов, то есть у узла максимум один родитель.

Визуализируются деревья обычно корнем вверх, а листьями вниз.
