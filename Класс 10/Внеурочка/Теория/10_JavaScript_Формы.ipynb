# Формы


Чтобы обработать данные нужно откуда-то эти данные считать, а чтобы их считать пользователю нужно эти данные куда-то ввести. Для этого существуют формы.


Для обозначения формы используется тег `<form>`. Внутри для группировки элементов формы может использоваться тег `<fieldset>`, а внутри группы можно установить заголовок с помощью тега `<legend>`
```html
<form>
    <fieldset>
        <legend>Заголовок группы</legend>
        ...
    </fieldset>
    <fieldset>
        ...
    </fieldset>
</form>  
```

## Элементы формы
### `input`
Для ввода данных используется тег `<input>` с разными атрибутами `type`, которые определяют тип данных для ввода. Тег `<input>` - одинарный.
```html
<input type="">
```
#### Общие атрибуты
|Атрибут|Значения|Описание|
|--|--|--|
| type | см. ниже | Тип данных для ввода|
| autocomplete | `on`/`off` | Включает/отключает автозаполнение браузером |
| name | строка | Имя поля, которое будет использоваться при отправке на сервер |
| value | соответствующее типу значение | Значение по умолчанию |
| disabled | само наличие атрибута | Элемент отключен для ввода |
| readonly | само наличие атрибута | Элемент отключен, но значение отправлено на сервер при отправке формы |
| required | само наличие атрибута | Элемент обязателен для заполнения |



#### Текстовые поля
##### Атрибуты
|Атрибут|Значения|Описание|
|--|--|--|
| placeholder | строка | Подсказывающий текст, который отображается в пустом поле |
| pattern | регулярное выражение | Регулярное выражение для валидации введенных данных |
| maxlength | число | Максимальное количество символов, которое можно ввести |
| minlength | число | Минимальное количество символов для валидного значения	 |
| size | число | Ширина текстового поля в символах	 |
##### Типы
Типы нужны не только для разного отображения и обработки данных, но и для корректного автозаполнения со стороны браузера. 
|Тип|Описание|
|--|--|
| text | Однострочное текстовое поле. Тип по умолчанию |
| password | text, но введеная строка показывается как звездочки |
| email | text, но для адреса электронной почты |
| tel | text, но для телефона |
| url | text, но для ссылки |
| search |  text, но для поиска |

Пример:
```html
<input type="text" id="text" name="text" placeholder="Введите текст" maxlength="50" required/>
```
<input type="text" id="text" name="text" placeholder="Введите текст" maxlength="50" required />

```html
<textarea name="message" rows="5" cols="30">
 Кот играл в саду.
</textarea>
```

<label><input type="radio" name="hobby" value="reading" />Чтение</label>

### Списки (`select`)
Выпадающие списки обзначаются отдельным тегом `<select>`. Для элементов списка используется тег `<option>`. Элементы могут объединяься в группы с помощью `<optgroup>`.
```html
<select>
  <optgroup label="">
    <option>...</option>
    <option>...</option>
  </optgroup>
  <optgroup label="">
    <option>...</option>
    <option>...</option>
  </optgroup>
</select>
```
Атрибуты `<select>`
|Атрибут|Значения|Описание|
|--|--|--|
| multiple | само наличие атрибута |  |
| size | число | Количество строк, которое видно за раз |

Атрибуты `<option>`:
|Атрибут|Значения|Описание|
|--|--|--|
| value | строка | Значение поля для отправки на сервер |
| selected | само наличие атрибута | Выбраный элемент |

```html
<label for="select">Простой список (select):</label>
<select id="select" name="city">
  <option value="">-- Выберите город --</option>
  <option value="msk">Москва</option>
  <option value="spb" selected>Санкт-Петербург</option>
  <option value="nsk">Новосибирск</option>
</select>
<br>
<label for="select-group">Список с группами (optgroup):</label>
<select id="select-group" name="country">
  <optgroup label="Европа">
    <option value="ru">Россия</option>
    <option value="de">Германия</option>
  </optgroup>
  <optgroup label="Азия">
    <option value="cn">Китай</option>
    <option value="jp">Япония</option>
  </optgroup>
</select>
<br>
<label for="multiple">Множественный выбор (multiple):</label>
<select id="multiple" name="languages" multiple size="4">
  <option value="python">Python</option>
  <option value="js">JavaScript</option>
  <option value="cpp">C++</option>
  <option value="java">Java</option>
</select>
```

## Отправка формы
Атрибуты `<form>`
|Атрибут|Значения|Описание|
|--|--|--|
| action | URL-адрес | URL на который отправляются данные формы |
| method | число | HTTP метод |

Для отправки формы используется событие `submit`, которое можно вызвать с помощью кнопки
```js
<button type="submit">Отправить форму (submit)</button>
```

В JavaScript обработка выглядит так:

```js
document.getElementById('myForm').addEventListener('submit', function(event) {
    event.preventDefault(); // Отменяет стандартную отправку формы
    
    // Собираем данные формы
    const formData = new FormData(this);
    
    // Отправляем через fetch
    fetch(this.action, {
        method: this.method,
        body: formData
    });
});
```
