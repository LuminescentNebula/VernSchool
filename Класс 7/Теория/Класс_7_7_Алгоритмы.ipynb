# Алгоритмы

**Алгоритм** — это точное описание порядка действий, которые должен выполнить исполнитель для решения задачи за конечное время.

Возвращаемся к тому, с чего начинали. 

Ввод, Вывод, Переменные, Вычисления, Ветвления и Циклы - самое основное из чего состоят все программы. Есть еще немало тем в программировании, но они в каком-то смысле только облегчают написание программ и можно обойтись и без них, при желании, но те темы, которые мы уже прошли должны позволить вам писать любые программы уже сейчас, *в теории*.

---

Иногда перед программистами стоит абстрактная задача и приходится сидеть, рисовать на листочке, думать как её решить, потому что какого-то общего принципа для решения этой задачи никто еще не придумал (не так часто она встречается допустим). А иногда для решения программы нужно не придумать новый алгоритм, а применить уже существующий.

Существует множество алгоритмов, которые уже описаны и их можно просто взять и использовать, например для алгоритмы для сортировки, для поиска, для быстрых вычислений. Один из таких алгоритмов был придуман еще задолго до компьютеров и программирования.

## Алгоритм Евклида (НОД)
Алгоритм Евклида — эффективный алгоритм для нахождения наибольшего общего делителя двух целых чисел. Был придуман еще в древней Греции, но до сих пор применим и используется во многих сферах, например в шифровании и кодировании, и конечно в математических вычислениях.

Самый очевидный способ нахождения НОД - разложить на простые множители каждое из чисел и перемножить общие из них между друг другом, но такой способ занимает много времени особенно при больших числах. Алгоритм Евклида предлагает другой способ, который не требует перебора всех множителей:

**Описание алгоритма**
1) Возьмите два числа A и B
2) Найдите остаток от деления большего числа на меньшее
3) Замените большее число на полученный остаток
4) Повторять шаги 2-3, пока остаток от деления не станет 0
5) Предыдущий остаток (число на которое делили) - НОД



Пример:
- A=48 B=18
- A=48 % 18
- A=12 B=18
- B=18 % 12
- A=12 B=6
- A=12 % 6=0
- НОД = 6


**Алгоритм в виде блок-схемы**

### НОК

НОК - Наименьшее общее кратное.

Для нахождения НОК можно воспользоваться тем же вариантом с разложением на множители, но это будет также долго. Другой вариант - нахождение НОК через НОД, то есть через Алгоритм Евклида.

Для этого просто надо перемножить изначальные числа A и B и разделить их произведение на НОД:

НОК(A, B) = (A*B) / НОД(A,B)
