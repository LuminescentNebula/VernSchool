# Функции в Python

## Что такое функция

Функция — это блок организованного кода, который можно использовать повторно для выполнения определенной задачи. Функции помогают структурировать программу, делают код более читаемым и избегают повторений.

В Python определяется с помощью ключевого слова `def`, за которым следует имя функции и круглые скобки с параметрами. После круглых скобок `:`, а за тем с новой строки с отступом код.

```python
def имя_функции(параметры):
    Код
    return результат
```

Чтобы вызвать функцию нужно написать `имя_функции(параметры)`:
```python
def greet():
    print("Привет, мир!")

# Вызов функции
greet()  # Выведет: Привет, мир!
```

Функции нужны для уменьшения дублирования кода. Обернув код в функцию вы сможете обращаться к ней где угодно и сколько угодно. Если функция делает слишком много действий, то лучше, разбить ее на несколько отдельных функций.

Для соединения нескольких слов в название функции принято использовать `_`. Для названий функций используйте глаголы (`calculate_sum`, `get_user`, `validate_input`).


## Параметры и аргументы
**Параметры** — это переменные, которые указываются при определении функции.   
**Аргументы** — это конкретные значения, передаваемые в функцию при ее вызове.
```python
def func(парметер):
    pass

func(аргумент)
```
### Позиционные параметры
Позиционные параметры передаются в функцию в порядке их объявления:
```python
def calculate_sum(a, b):
    sum_result = a + b
    return sum_result

print(calculate_sum(2, 3))
```

### Именованные (ключевые) аргументы
Именованные аргументы можно указывать в произвольном порядке.
```python
def display_info(name, age, profession):
    print(f"{name} — {profession}, {age} лет")

# Вызов с именованными аргументами
display_info(profession="программист", name="Максим", age=30)
# Выведет: Максим — программист, 30 лет
```

### Параметры со значениями по умолчанию
Можно задать значения по умолчанию для параметров. В этом случае при вызове их можно будет не указывать
```python
def func(a, b, c=2):
    return a + b + c

print(func(1, 2))     # 5 (a=1, b=2, c=2 по умолчанию)
print(func(1, 2, 10)) # 13 (a=1, b=2, c=10)
```

### Переменное число аргументов
#### *args — произвольное число позиционных аргументов
Внутри функции `args` будет кортежем - неизменяемым списком.
```python
def sum_all(*args):
    total = 0
    for num in args:
        total += num
    return total

print(sum_all(1, 2, 3))        # 6
print(sum_all(10, 20, 30, 40)) # 100
```
`[1,2,3,4]` - массив, 1 аргумент  
`1,2,3,4` - 4 аргумента

#### **kwargs — произвольное число именованных аргументов
```python
def print_info(**kwargs):
    for key, value in kwargs.items():
    print(f"{key}: {value}")

print_info(name="Анна", age=25, city="Москва")
# Выведет:
# name: Анна
# age: 25
# city: Москва
```


## Возврат значений

**Функция** - возвращает значение  
**Процедура** - ничего не возвращает


В Python и функция и процедура обозначаются одинаково через `def`, разница только в наличии возвращаемого значения.

Функция может возвращать значения с помощью оператора `return`:
```python

def multiply(x, y):
    return x * y

result = multiply(5, 3)
print(result)  # 15
```

Можно возвращать несколько значений в виде кортежа:

```python

def get_min_max(numbers):
    return min(numbers), max(numbers)

minimum, maximum = get_min_max()
print(f"Минимум: {minimum}, Максимум: {maximum}")
# Выведет: Минимум: 1, Максимум: 9

```


## Lambda-функции

Lambda-функции — это анонимные функции, определяемые в одной строке:

```python
# Обычная функция
def square(x):
    return x ** 2

# Lambda-функция
square_lambda = lambda x: x ** 2

print(square(5))        # 25
print(square_lambda(5)) # 25

```

`lambda` часто используется с функциями `map()`, `filter()`. Обе эти функции принимают в качестве аргументов фунцию и список значений. Значения передаются функцию по одному. 

`map` применяет функцию к каждому значению списка и возвращет новую последовательность из возвращенных значений функций.

```python
numbers = [1,2,3,4,5]
# Умножить каждое число на 2
doubled = list(map(lambda x: x * 2, numbers))
print(doubled) 
# [2, 4, 6, 8, 10]
```

`filter` применяет функцию к каждому значению списка и возвращет новую последовательность, в которой будут только значения, для которых функция вернула `True`. 

```python
# Отфильтровать четные числа
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
#[2, 4]
```


## Области видимости
### Локальные и глобальные переменные
Переменные, которые были объявлены вне любых функций - **глобальные**.

Переменные, которые были объявлены внутри функции - **локальные**.

К глобальным можно обратиться откуда угодно, а локальные доступны только  пределах своей функции и удаляются в конце её выполнения.

**Функция не должна изменять глобальные переменные без явной необходимости.**

``` python
x = 10  # Глобальная переменная

def test_scope():
    x = 5  # Локальная переменная
    print(f"Внутри функции: x = {x}")

test_scope()  # Выведет: Внутри функции: x = 5
print(f"Вне функции: x = {x}")  # Выведет: Вне функции: x = 10
```

### Использование global
Чтобы объявить глобальную переменную внутри функции можно использовать ключевое слово `global`.
```python
counter = 0

def increment():
    global counter
    counter += 1

increment()
increment()
print(counter)  # 2
```



## Функции как объекты

В Python функции являются объектами — их можно передавать как аргументы и возвращать из других функций:

```python
def create_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier

#Записываем функцию в переменную
times_three = create_multiplier(3)
times_five = create_multiplier(5)

print(times_three(10))  # 30
print(times_five(10))   # 50
```
